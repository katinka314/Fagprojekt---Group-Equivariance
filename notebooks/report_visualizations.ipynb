```python
import numpy as np
import matplotlib.pyplot as plt

# Domain
x = np.linspace(-3, 6.2, 1000)

# Gaussian function
def gauss(x, mu, sigma=0.7):
    return np.exp(-(x - mu)**2 / (2 * sigma**2))

# Curves
y1 = gauss(x, 0)
y2 = gauss(x, 4)

fig, ax = plt.subplots(figsize=(10, 4))

# First Gaussian
ax.plot(x, y1, color="#5a8f52", lw=2.5)

# Second Gaussian
ax.plot(x, y2, color="#cc554d", lw=2.5)

# Peak markers
ax.scatter([0, 4], [1, 1],
           s=70, color="slateblue",
           edgecolors="black", zorder=5)

# x-axis markers
ax.scatter([0, 4], [0, 0],
           s=70, color="slateblue",
           edgecolors="black", zorder=5)

ax.text(-0.3, 0.1, r"$g^{-1}x'$", color="#7d7dff", fontsize=20)
ax.text(3.95, 0.1, r"$x'$", color="#7d7dff", fontsize=20)

# Small labels
ax.text(-0.2, 1.1, r"$f(x)$", color="#5a8f52", fontsize=18)
ax.text(3.2, 1.1, r"$L_gf(x) = f(g^{-1}x)$", color="#cc554d", fontsize=18)

# Axes through origin
ax.spines["left"].set_position("zero")
ax.spines["bottom"].set_position("zero")
ax.spines["left"].set_visible(False)
ax.tick_params(axis="y", left=False, labelleft=False)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)

# Limits
ax.set_xlim(-3.1, 6.1)
ax.set_ylim(-0.05, 1.25)

plt.tight_layout()
plt.show()
```

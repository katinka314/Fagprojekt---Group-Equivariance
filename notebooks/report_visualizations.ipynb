```python
import numpy as np
import matplotlib.pyplot as plt

# Domain
x = np.linspace(-3, 6.2, 1000)

# Gaussian function
def gauss(x, mu, sigma=0.7):
    return np.exp(-(x - mu)**2 / (2 * sigma**2))

# Curves
y1 = gauss(x, 0)
y2 = gauss(x, 4)

fig, ax = plt.subplots(figsize=(10, 4))

# First Gaussian
ax.plot(x, y1, color="#5a8f52", lw=2.5)

# Second Gaussian
ax.plot(x, y2, color="#cc554d", lw=2.5)

# Peak markers
ax.scatter([0, 4], [1, 1],
           s=70, color="slateblue",
           edgecolors="black", zorder=5)

# x-axis markers
ax.scatter([0, 4], [0, 0],
           s=70, color="slateblue",
           edgecolors="black", zorder=5)

ax.text(-0.3, 0.1, r"$g^{-1}x'$", color="#7d7dff", fontsize=20)
ax.text(3.95, 0.1, r"$x'$", color="#7d7dff", fontsize=20)

# Small labels
ax.text(-0.2, 1.1, r"$f(x)$", color="#5a8f52", fontsize=18)
ax.text(3.2, 1.1, r"$L_gf(x) = f(g^{-1}x)$", color="#cc554d", fontsize=18)

# Axes through origin
ax.spines["left"].set_position("zero")
ax.spines["bottom"].set_position("zero")
ax.spines["left"].set_visible(False)
ax.tick_params(axis="y", left=False, labelleft=False)
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)

# Limits
ax.set_xlim(-3.1, 6.1)
ax.set_ylim(-0.05, 1.25)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6,6))

# Circle of radius 2
circle = plt.Circle((0, 0), 2, fill=False, color='gray', linewidth=2)
ax.add_patch(circle)

# Axes
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)

# Points (angles in radians)
theta_red = np.pi/2      # 90 degrees
theta_blue = 2*np.pi/3   # 120 degrees

r = 2

# Vectors
red = np.array([r*np.cos(theta_red), r*np.sin(theta_blue)])
blue = np.array([r*np.cos(theta_blue), r*np.sin(theta_blue)])

ax.arrow(0, 0, red[0], red[1], head_width=0.08, length_includes_head=True, color='red', linewidth=2)
ax.arrow(0, 0, blue[0], blue[1], head_width=0.08, length_includes_head=True, color='blue', linewidth=2)

# Angle arcs
arc_r = 0.5

t1 = np.linspace(0, theta_red, 100)
ax.plot(arc_r*np.cos(t1), arc_r*np.sin(t1), color='red')

t2 = np.linspace(0, theta_blue, 100)
ax.plot(0.7*np.cos(t2), 0.7*np.sin(t2), color='blue')

# Labels
ax.text(0.1, 0.2, r'$\theta$', color='red', fontsize=14)
ax.text(0.4, 0.35, r'$\theta$', color='blue', fontsize=14)

# Formatting
ax.set_xlim(-2.2, 2.2)
ax.set_ylim(-2.2, 2.2)
ax.set_aspect('equal')
ax.set_xticks(range(-2, 3))
ax.set_yticks(range(-2, 3))
ax.set_yticklabels([r'$-2i$', r'$-i$', r'$0$', r'$i$', r'$2i$'])
ax.grid(True, alpha=0.3)

plt.show()
```
